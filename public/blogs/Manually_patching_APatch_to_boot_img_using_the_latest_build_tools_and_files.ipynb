## 🎈 Setting up environment and installing essential tools for Magickboot

````python
%%shell
sudo apt-get -y install build-essential && \
git clone -c core.symlinks=true https://github.com/topjohnwu/Magisk.git && \
cd Magisk && \
git submodule update --init --recursive
````

## 🚡 Building Magickboot latest

````python
%%shell
mkdir -p android_sdk && \
cd Magisk && \
ANDROID_HOME=/content/android_sdk python build.py -v ndk && \
ANDROID_HOME=/content/android_sdk python build.py -v native
````

## 🐤 Downloading essential files for APatch patch boot.img
Don't forget to check latest KernelPatch file for latest build
https://github.com/bmax121/KernelPatch/releases

````python
%%shell
wget -O kpimg-android https://github.com/bmax121/KernelPatch/releases/download/0.11.1-dev/kpimg-android
wget -O kptools-linux https://github.com/bmax121/KernelPatch/releases/download/0.11.1-dev/kptools-linux
chmod +x kpimg-android kptools-linux
````

> ## ❗❗❗ Upload your boot.img to Colab files directory /content/




## 👽 Finaliy patching boot.img with APatch

````python
%%shell
rm -f ramdisk.cpio kernel-b kernel new-boot.img && \
/content/Magisk/native/out/x86_64/magiskboot unpack boot.img && \
mv kernel kernel-b && \
./kptools-linux -p --image kernel-b --skey "Meme1234" --kpimg kpimg-android --out kernel && \
/content/Magisk/native/out/x86_64/magiskboot repack boot.img
````

## ✅ If all command run successfully then now you just download the new-boot.img and flash to your phone via fastboot command

```
fastboot flash boot_a new-boot.img
fastboot flash boot_b new-boot.img
```

And Super key password: `Meme1234`
